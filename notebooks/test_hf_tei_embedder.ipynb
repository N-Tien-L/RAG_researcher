# HuggingFace TEI Embedder Smoke Test
Run the cells below to send sample passages to your Hugging Face text-embeddings-inference server. 

```python
import os
from app.embeddings.huggingface_tei import HuggingFaceTEIEmbedder

TEI_BASE_URL = os.getenv('TEI_URL', 'http://localhost:8080')
MAX_BATCH_SIZE = int(os.getenv('TEI_MAX_BATCH', '8'))
MODE = os.getenv('TEI_MODE', 'passage')

print(f'Using TEI server at: {TEI_BASE_URL}')
```

```python
sample_texts = [
    "RAG pipelines benefit from semantic chunk embeddings.",
    "You can switch the TEI mode to query when embedding user questions.",
]

embedder = HuggingFaceTEIEmbedder(
    base_url=TEI_BASE_URL,
    max_batch_size=MAX_BATCH_SIZE,
    mode=MODE,
)

embeddings = embedder.embed_documents(sample_texts)
print(f'Generated {len(embeddings)} embeddings; first vector length: {len(embeddings[0])}')
embeddings[0][:8]  # peek at first 8 dims
```
