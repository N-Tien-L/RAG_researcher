# RagPipeline Ingestion Test
This notebook wires up `RagPipeline.ingest_data()` so you can push a PDF path or YouTube URL into Chroma.

```python
import os
from datetime import datetime, timezone

from app.rag.pipeline import RagPipeline

SOURCE_TYPE = "youtube"
PDF_PATH = os.getenv("INGEST_PDF_PATH", "C:/Users/Tien/OneDrive/Documents/Tien_Lam_CV.pdf")
YOUTUBE_URL = os.getenv("INGEST_YOUTUBE_URL", "https://www.youtube.com/watch?v=O2UmHpNlwUw")
COLLECTION_NAME = os.getenv("INGEST_COLLECTION", "notebook_test")

pipeline = RagPipeline(max_tokens=300, overlap=40)

if SOURCE_TYPE == "pdf":
    source_value = PDF_PATH
elif SOURCE_TYPE == "youtube":
    source_value = YOUTUBE_URL
else:
    raise ValueError("SOURCE_TYPE must be 'pdf' or 'youtube'")

extra_metadata = {
    "ingested_from": "notebook",
    "notebook_run_ts": datetime.now(timezone.utc).isoformat() + "Z",
}

print(f"Configured ingestion -> type: {SOURCE_TYPE}, source: {source_value}")
print(f"Collection: {COLLECTION_NAME}")
```

```python
try:
    result = pipeline.ingest_data(
        source=source_value,
        collection_name=COLLECTION_NAME,
        source_type=SOURCE_TYPE
    )
    print("Ingestion complete.")
    print(result)
except Exception as exc:
    print(f"Ingestion failed: {exc}")
    raise
```
